```python
# --- Streaming Example ---

stream_messages = [
    {"role": "system", "content": "You are a storyteller. Tell vivid, short tales."},
    {"role": "user", "content": "Tell me a two-paragraph story about a robot who discovers music."}
]

pretty_print("Streaming response:\n")

stream = client.chat.completions.create(
    model="gpt-5-nano",
    messages=stream_messages,
    stream=True
)

collected_reply = []
for chunk in stream:
    token = chunk.choices[0].delta.content
    if token is not None:
        print(token, end="", flush=True)
        collected_reply.append(token)

pretty_print("\n\n--- stream complete ---")
```

```python
import time
from openai import (
    AuthenticationError,
    RateLimitError,
    BadRequestError,
    APIConnectionError,
    APITimeoutError,
)


def safe_chat(client, messages, model="gpt-5-nano", **kwargs):
    """Wrapper that catches common API errors gracefully."""
    try:
        return client.chat.completions.create(
            model=model,
            messages=messages,
            **kwargs
        )
    except AuthenticationError:
        pretty_print("Authentication failed - check your API key.")
    except RateLimitError:
        pretty_print("Rate limit hit - waiting 5s then you can retry.")
        time.sleep(5)
    except BadRequestError as e:
        pretty_print(f"Bad request: {e.message}")
    except APIConnectionError:
        pretty_print("Network error - check your internet connection.")
    except APITimeoutError:
        pretty_print("Request timed out - try again.")
    except Exception as e:
        pretty_print(f"Unexpected error: {e}")
    return None


# Quick test
resp = safe_chat(client, [{"role": "user", "content": "Say hello in one word."}])
if resp:
    print("safe_chat works:", resp.choices[0].message.content)
```

## Response Id

```python
# Turn 1: establish persistent behavior in the thread
r1 = client.responses.create(
    model="gpt-5-nano",
    input=[
        {"role": "developer", "content": "You are a grumpy pirate code reviewer. Be sarcastic. And answer in as few words as possible."},
        {"role": "user", "content": "Explain recursion."},
    ],
)


print("Turn 1 response:", r1.output_text)
```

```python
# Turn 2: same thread, but for THIS answer force JSON only (one-off)
r2 = client.responses.create(
    model="gpt-5-nano",
    previous_response_id=r1.id,
    instructions="Output valid JSON ONLY with keys: definition, example.",
    input=[{"role": "user", "content": "Now give me an example in Python."}],
)

print("Turn 2 response:", r2.output_text)
```

```python

r3 = client.responses.create(
    model="gpt-5-nano",
    instructions="Output valid JSON ONLY with keys: definition, example.",
    input=[{"role": "user", "content": "Now give me an example in Python."}],
    store=False
)

print("Turn 3 response:", r3.output_text)
```

```python
# Turn 4: wont be able to refer to r3 because it was not stored.
r4 = client.responses.create(
    model="gpt-5-nano",
    previous_response_id=r3.id,
    instructions="Output valid JSON ONLY with keys: definition, example.",
    input=[{"role": "user", "content": "Now give me an example in Python."}],

)

print("Turn 4 response:", r4.output_text)
```

# When to use instructions, when to use role developer, and when to use system prompts

```python
# Turn 1: an internal triage note the backend would store in Zendesk as JSON
r1 = client.responses.create(
    model="gpt-5-nano",
    instructions=(
        "You are an internal support triage bot. Return only valid JSON with keys "
        "severity, suspected_causes, next_questions and do not write any customer-facing text."
    ),
    input=(
        "A customer reports webhook deliveries started retrying heavily since 10:42 UTC "
        "and they see 502 errors from our endpoint on the Pro plan."
    ),

    #reasoning={"effort": "high"},    
	reasoning = {"effort": "minimal"}, # Efforts for reasoning are very minimal, so it won't end up overthinking a lot. The model 
	text={"verbosity": "low"}, # as low words as possible, but still answer the question
)

print("TURN 1:\n", r1.output_text)
```

```python
# Turn 2: continue the same thread, but do NOT pass instructions again
r2 = client.responses.create(
    model="gpt-5-nano",
    previous_response_id=r1.id,
    input=(
        "Now write a customer-facing email reply: apologize, explain what we're checking, "
        "and ask for 2 specific details. Plain English, not JSON."
    ),
    reasoning = {"effort": "minimal"},
    text={"verbosity": "low"},
)

print("\nTURN 2:\n", r2.output_text)

# Now, even though the instructions from the previous response R1.id were saying that we need 
# an output only in JSON format, the input in the current API call in R2 says 
# that it should be in plain English, not JSON. 
# The instructions from the previous API call are not getting carried forward to this
#  current API call. 
# And because of this reason, the instructions from the previous API call do not have an effect 
# on this API call, and whatever we have mentioned in our input will be followed. 
```

```python
r1 = client.responses.create(
    model="gpt-5-nano",
    input=[
        {"role": "developer", "content": (
            "You are an internal support triage assistant and you must always output only valid "
            "JSON with keys severity, suspected_causes, next_questions and never produce "
            "customer-facing prose."
        )},
        {"role": "user", "content": (
            "Customer reports webhook deliveries started retrying heavily since 10:42 UTC "
            "and they see 502 errors from our endpoint on the Pro plan."
        )}
    ],
    reasoning = {"effort": "minimal"},
    text={"verbosity": "low"},
)

print("TURN 1:\n", r1.output_text)
```

```python
# Exactly the same turn-2 request as before
r2 = client.responses.create(
    model="gpt-5-nano",
    previous_response_id=r1.id,
    input="Now write a customer-facing email apology explaining what we're checking and ask for exactly two specific details in plain English.",
    reasoning={"effort": "minimal"},
    text={"verbosity": "low"}
)

print("\nTURN 2:\n", r2.output_text)

# Now, in this example, we used the developer role to pass the instructions to carry over the cells 
# and the API calls. The R1 request contained the developer role, and we are passing that R1.id 
# as a previous response ID in the R2 request. 
# Because of that, the R2 request also follows the same instructions as those provided in the 
# developer role. 
# Here in R2 as well, the model is following the same instructions as of in r1 developer role content. 
# Because of that, even though our input says to give the details or send the mail in plain English, 
# the model still generates the output in JSON format. 
```

```python
# Surface 1: Slack. Policy in `developer`, format in `instructions`.
r1 = client.responses.create(
    model="gpt-5-nano",
    input=[
        {"role": "developer", "content": (
            "You are ACME Incident Assistant; do not guess unknown facts; if unsure say what you "
            "need; keep recommendations actionable; do not expose internal-only details."
        )},
        {"role": "user", "content": (
            "We're seeing intermittent payment failures in EU; gateway 502 spike started "
            "08:12 UTC; failover reduced it but not fully."
        )},
    ],
    instructions="Output as a Slack message with at most 6 lines and include one short checklist.",
)

print(r1.output_text)
```

```python
# Surface 2: Jira. Same thread, brand-new output contract.
r2 = client.responses.create(
    model="gpt-5-nano",
    previous_response_id=r1.id,
    input="Convert this into a Jira incident update.",
    instructions="Return only JSON with keys summary, customer_impact, current_status, next_actions.",
)

print(r2.output_text)
```
